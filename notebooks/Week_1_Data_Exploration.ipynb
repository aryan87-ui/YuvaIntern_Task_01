# Week 1 — Data Exploration and Problem Definition

## Objective
Explore the Sample Superstore dataset, understand its variables and data quality, perform basic EDA, visualize important patterns, and define a business problem for further analysis.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('../data/Sample - Superstore.csv', encoding='latin1')
df['Order Date'] = pd.to_datetime(df['Order Date'], errors='coerce')
df['Ship Date'] = pd.to_datetime(df['Ship Date'], errors='coerce')
df.head()
```

## 1. Dataset Shape and Structure

```python
print('Rows, Columns:', df.shape)
print('\nColumns:')
print(df.columns.tolist())
print('\nData Types:')
print(df.dtypes)
```

## 2. Data Quality Checks

```python
print('Missing values:')
display(df.isnull().sum().to_frame('Missing Values'))

print('Duplicate rows:', df.duplicated().sum())
```

## 3. Summary Statistics

```python
df.describe(include='all').T
```

## 4. Key Business Metrics

```python
total_sales = df['Sales'].sum()
total_profit = df['Profit'].sum()
total_quantity = df['Quantity'].sum()
profit_margin = total_profit / total_sales * 100

print(f'Total Sales: {total_sales:,.2f}')
print(f'Total Profit: {total_profit:,.2f}')
print(f'Total Quantity: {total_quantity:,.0f}')
print(f'Profit Margin: {profit_margin:.2f}%')
```

## 5. Sales and Profit by Category

```python
category_summary = df.groupby('Category')[['Sales', 'Profit']].sum().sort_values('Sales', ascending=False)
category_summary
```

```python
category_summary['Sales'].sort_values().plot(kind='barh', figsize=(8,5))
plt.title('Sales by Category')
plt.xlabel('Sales')
plt.tight_layout()
plt.show()
```

## 6. Regional Analysis

```python
region_summary = df.groupby('Region')[['Sales', 'Profit']].sum().sort_values('Sales', ascending=False)
region_summary
```

## 7. Monthly Sales Trend

```python
monthly_sales = df.set_index('Order Date')['Sales'].resample('MS').sum()
monthly_sales.plot(figsize=(10,5))
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.tight_layout()
plt.show()
```

## 8. Initial Findings

- Sales and profit should be analyzed together.
- Negative-profit transactions require investigation rather than automatic deletion.
- Category, region, product, customer segment, discount and time are important analytical dimensions.
- Monthly sales can be used to study trend and seasonality.

## 9. Problem Definition

The business needs to understand the factors associated with sales and profitability across products, categories, regions, customer segments and time.

### Key Questions
1. How do sales and profit change over time?
2. Which products/categories generate the most sales and profit?
3. Which products have high sales but low or negative profit?
4. Does discounting appear to reduce profitability?
5. Which regions and customer segments contribute most to revenue and profit?

## 10. Hypotheses
- H1: Higher discounts are associated with lower profit margins.
- H2: Sales and profit differ across categories/sub-categories.
- H3: High-sales products are not necessarily high-profit products.
- H4: Regional performance differs in revenue and profitability.
- H5: Sales show time-based patterns or seasonality.
